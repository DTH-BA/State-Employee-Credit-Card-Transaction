```python
import numpy as np

# Ensure date type
for c in ['TRANS_DT','TX_DATE']:
    if c in df.columns:
        df[c] = pd.to_datetime(df[c], errors='coerce')

# Helper function
def is_binary(s: pd.Series) -> bool:
    vals = pd.Series(s.dropna().unique())
    if len(vals) == 0: 
        return False
    # accept {0,1} or {True,False}
    return set(pd.Series(vals).astype(str)).issubset({'0','1','True','False'})

# Identify column groups
datetime_cols = [c for c in df.columns if np.issubdtype(df[c].dtype, np.datetime64)]
numeric_cols  = [c for c in df.columns if np.issubdtype(df[c].dtype, np.number)]
object_cols   = [c for c in df.columns if df[c].dtype == 'object']

# Binary flags (prefer names starting with IS_ or values ​​0/1)
flag_cols_by_name = [c for c in df.columns if c.upper().startswith('IS_')]
flag_cols_by_val  = [c for c in numeric_cols if is_binary(df[c])]
flag_cols = sorted(list(set(flag_cols_by_name + flag_cols_by_val)))

# Discrete Numeric (like FISCAL_YEAR/PERIOD, TX_YEAR/MONTH, KEY_GROUP_SIZE)
# criteria: number of unique values ​​<= 100 (adjustable) and NOT binary
discrete_numeric_cols = [
    c for c in numeric_cols 
    if c not in flag_cols and df[c].nunique(dropna=True) <= 100
]

# Continuous Numeric (amount, etc.)
continuous_numeric_cols = [
    c for c in numeric_cols 
    if c not in flag_cols and c not in discrete_numeric_cols
]

# Categorical (object + discrete numerics with categorical meaning)
categorical_cols = object_cols.copy()

summary_types = {
    "datetime_cols": datetime_cols,
    "flag_cols": flag_cols,
    "discrete_numeric_cols": discrete_numeric_cols,
    "continuous_numeric_cols": continuous_numeric_cols,
    "categorical_cols": categorical_cols,
}

for k,v in summary_types.items():
    print(f"{k} ({len(v)}): {v}")
```

## Log-Histogram of AMOUNT_ABS
### Objective
The objective of this step is to observe the distribution of the variable `AMOUNT_ABS` after applying the logarithm transformation (log1p).
The reason for using log-transformation is because the original data of `AMOUNT_ABS` has a very large dispersion, many small values but also some very large values (outliers). This makes the original histogram right-skewed and difficult to analyze.

The log-transform helps to:
* Shrink the gap between large and small values.
* Reduce skewness, making the distribution closer to normal distribution.
* Easily observe statistical characteristics (mean, median, mode).

```python
import numpy as np

# Histogram log scale for AMOUNT_ABS
plt.figure(figsize=(8,5))
plt.hist(np.log1p(df['AMOUNT_ABS']), bins=50, color='salmon', edgecolor='black')
plt.title("Log-Histogram of AMOUNT_ABS")
plt.xlabel("log(1 + AMOUNT_ABS)")
plt.ylabel("Frequency")
plt.show()
```

### Observation results
* After log-transform, the distribution of AMOUNT_ABS becomes more symmetrical.
* The majority of transactions are concentrated in the log(1+AMOUNT_ABS) range from 3 to 6.
* A very large number of transactions (tens of thousands) fall into this range, indicating that small and medium-sized transactions are the majority.
* A few transactions still have very high log values ​​(8–12), indicating rare but extremely large value transactions (outliers).

## Log-Histogram of MERCHANDISE_AMT
### Objective
In this step, we want to analyze the `MERCHANDISE_AMT` variable further by dividing it into 2 groups:
* Positive values (revenue) – successful sales transactions.
* Negative values (refunds) – refund transactions (customers return).

This division helps to compare the distribution between actual revenue and refunds, thereby assessing the nature of commercial activities.

```python
# Divide data into positive and negative
positive_amt = df[df['MERCHANDISE_AMT'] > 0]['MERCHANDISE_AMT']
negative_amt = df[df['MERCHANDISE_AMT'] < 0]['MERCHANDISE_AMT']

plt.figure(figsize=(12,5))

# Histogram for positive
plt.subplot(1,2,1)
plt.hist(np.log1p(positive_amt), bins=50, color='green', edgecolor='black')
plt.title("Log-Histogram of Positive MERCHANDISE_AMT")
plt.xlabel("log(1 + MERCHANDISE_AMT)")
plt.ylabel("Frequency")

# Histogram for negative
plt.subplot(1,2,2)
plt.hist(np.log1p(-negative_amt), bins=50, color='red', edgecolor='black')  
plt.title("Log-Histogram of Negative MERCHANDISE_AMT (Refunds)")
plt.xlabel("log(1 + |MERCHANDISE_AMT|)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## Observation results
### Positive Transactions (Blue)
* The log-transformed distribution is almost normal.
* Common transactions are in the range log(1+AMT) = 3–6, meaning transactions are of average value – not too small and not too large.
* The number of transactions is very large (tens of thousands), indicating that sales activities dominate the dataset.
### Negative Transactions – Refunds (Red)
* The distribution is also almost normal but has a much smaller number (maximum frequency is only ~1000 compared to 40,000 for positive).
* This shows that refunds are rare events, accounting for only a small fraction of all transactions.
* However, there are also some large-value refunds, as shown by the right tail of the histogram.

```python
# List of flag variables
flag_vars = [
    'IS_REFUND', 'IS_HIGH_VALUE', 'IS_EXACT_DUP', 'IS_WEEKEND',
    'IS_UNKNOWN_DEPT_NAME', 'IS_UNKNOWN_DIV_NAME',
    'IS_UNKNOWN_MERCHANT', 'IS_UNKNOWN_CAT_DESCR'
]

import matplotlib.pyplot as plt

for col in flag_vars:
    counts = df[col].value_counts(dropna=False)
    total = counts.sum()
    
    plt.figure(figsize=(5,4))
    counts.plot(kind='bar', color=['skyblue','salmon'], edgecolor='black')
    plt.title(f"Distribution of {col}")
    plt.xticks(rotation=0)
    plt.xlabel("Flag Value (0=No, 1=Yes)")
    plt.ylabel("Count")
    
    # Show the percentage (%)
    for i, v in enumerate(counts):
        plt.text(i, v + total*0.005, f"{v/total:.2%}", ha='center')
    
    plt.show()
```

## Flag Variables
### IS_REFUND
* The percentage of refund transactions (Yes = 1) is ~2%, while the majority (98%) are not refunded.
* Refunds exist but are rare.
### IS_HIGH_VALUE
* High value transactions (Yes) account for 0.51%, which is extremely low compared to the entire dataset.
* Most transactions are normal value.
### IS_EXACT_DUP
* About 10.33% of transactions are marked as duplicates.
* This is not a small amount, much higher than rare flags (eg: refund 2%, high value 0.31%).
* This shows that duplicates appear relatively frequently in the data, which should be noted in the next analysis step.
### IS_WEEKEND

* Most transactions occur on weekdays.

## Flags for missing information (Unknown)
* The variables IS_UNKNOWN_DEPT_NAME, IS_UNKNOWN_DIV_NAME, IS_UNKNOWN_MERCHANT, IS_UNKNOWN_CAT_DESCR are all 0 for the entire data.
* This means that there are no cases where information about the department, division, merchant or category description is missing.
